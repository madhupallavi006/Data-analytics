```python
import pandas as pd
import zipfile
import matplotlib.pyplot as plt

zip_name = next(iter(uploaded))

with zipfile.ZipFile(zip_name, 'r') as zip_ref:
    zip_ref.extractall()
    print(zip_ref.namelist())

df = pd.read_csv('creditcard.csv')

print(df.head())

print(df.info())

print(df.isnull().sum())

df = df.drop_duplicates()

fraud_count = df['Class'].value_counts()

plt.figure(figsize=(6,4))

fraud_count.plot(kind='bar')

plt.title("Fraud vs Normal Transactions")

plt.xlabel("Class")

plt.ylabel("Count")

plt.show()

amount = df.groupby('Class')['Amount'].mean()

plt.figure(figsize=(6,4))

amount.plot(kind='bar')

plt.title("Average Transaction Amount")

plt.xlabel("Class")

plt.ylabel("Amount")

plt.show()

time = df.groupby('Class')['Time'].mean()

plt.figure(figsize=(6,4))

time.plot(kind='bar')

plt.title("Average Transaction Time")

plt.xlabel("Class")

plt.ylabel("Time")

plt.show()

top_amount = df.sort_values(by='Amount', ascending=False).head(10)

plt.figure(figsize=(8,5))

plt.bar(top_amount.index, top_amount['Amount'])

plt.title("Top 10 Transaction Amounts")

plt.xlabel("Transaction")

plt.ylabel("Amount")

plt.show()

print("\nFraud Transactions:", fraud_count[1])

print("Normal Transactions:", fraud_count[0])

print("\nBusiness Insights")

print("------------------")

print("Fraud transactions are much lower than normal transactions.")

print("Banks can use Machine Learning for fraud prevention.")

print("Large transaction amounts should be monitored carefully.")

print("Real-time fraud detection improves financial security.")
```
