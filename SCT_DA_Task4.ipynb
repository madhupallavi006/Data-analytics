```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
import zipfile

uploaded = files.upload()

zip_name = next(iter(uploaded))

with zipfile.ZipFile(zip_name, 'r') as zip_ref:
    zip_ref.extractall()

df = pd.read_csv("marketing_campaign.csv", sep=';')

print(df.head())

print(df.columns)

print(df.info())

print(df.isnull().sum())

df = df.drop_duplicates()

income = df.groupby('Education')['Income'].mean()

plt.figure(figsize=(8,5))

income.plot(kind='bar')

plt.title("Average Income by Education")

plt.xlabel("Education")

plt.ylabel("Income")

plt.show()

web = df.groupby('Education')['NumWebPurchases'].sum()

plt.figure(figsize=(8,5))

web.plot(kind='bar')

plt.title("Web Purchases by Education")

plt.xlabel("Education")

plt.ylabel("Purchases")

plt.show()

spent = df[['MntWines','MntFruits','MntMeatProducts',
            'MntFishProducts','MntSweetProducts']].sum()

plt.figure(figsize=(7,7))

spent.plot(kind='pie', autopct='%1.1f%%')

plt.title("Product Spending Distribution")

plt.ylabel('')

plt.show()

best = spent.idxmax()

print("Highest Spending Category:", best)

print("\nBusiness Insight Report")
print("-----------------------")
print("Customers spend most on:", best)
print("Focus marketing budget on this category.")
```
