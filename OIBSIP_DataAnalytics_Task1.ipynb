```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()

df = pd.read_csv(next(iter(uploaded)))

print(df.head())

print(df.columns)

print(df.info())

print(df.isnull().sum())

df = df.drop_duplicates()

df['Order Date'] = pd.to_datetime(df['Order Date'])

print(df.describe())

sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(8,5))

sales.plot(kind='bar')

plt.title("Sales by Category")

plt.xlabel("Category")

plt.ylabel("Sales")

plt.show()

profit = df.groupby('Region')['Profit'].sum()

plt.figure(figsize=(7,7))

profit.plot(kind='pie', autopct='%1.1f%%')

plt.title("Profit by Region")

plt.ylabel('')

plt.show()

monthly = df.groupby(df['Order Date'].dt.month)['Sales'].sum()

plt.figure(figsize=(8,5))

monthly.plot(kind='line', marker='o')

plt.title("Monthly Sales Trend")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.grid()

plt.show()

top = df.groupby(df.columns[2])['Sales'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(8,5))

top.plot(kind='barh')

plt.title("Top Selling Products")

plt.xlabel("Sales")

plt.show()

best = top.idxmax()

print("\nBusiness Insights Report")
print("-------------------------")
print("Highest Selling Product:", best)
print("Technology and Office Supplies show strong sales performance.")
print("Recommend focusing marketing budget on high-performing products.")
```
