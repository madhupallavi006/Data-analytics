```python
import pandas as pd
import matplotlib.pyplot as plt

# Read Dataset
df = pd.read_csv("SampleSuperstore.csv")

# Show Data
print(df.head())

# Remove Duplicates
df = df.drop_duplicates()

# Convert Date
df['Order Date'] = pd.to_datetime(df['Order Date'])

# -----------------------------
# TOTAL SALES & PROFIT
# -----------------------------

print("Total Sales:", df['Sales'].sum())
print("Total Profit:", df['Profit'].sum())

# -----------------------------
# SALES BY CATEGORY
# -----------------------------

sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(7,5))
sales.plot(kind='bar')

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.show()

# -----------------------------
# PROFIT BY REGION
# -----------------------------

profit = df.groupby('Region')['Profit'].sum()

plt.figure(figsize=(7,7))
profit.plot(kind='pie', autopct='%1.1f%%')

plt.title("Profit by Region")

plt.ylabel("")

plt.show()

# -----------------------------
# MONTHLY SALES TREND
# -----------------------------

monthly = df.groupby(df['Order Date'].dt.month)['Sales'].sum()

plt.figure(figsize=(8,5))
monthly.plot(kind='line', marker='o')

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid()

plt.show()

# -----------------------------
# TOP PRODUCTS
# -----------------------------

top = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(8,5))
top.plot(kind='barh')

plt.title("Top 10 Products")

plt.xlabel("Sales")

plt.show()
```
