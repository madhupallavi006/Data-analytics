```python
!pip install pandas plotly seaborn matplotlib -q

import pandas as pd
import plotly.express as px
import seaborn as sns
import matplotlib.pyplot as plt

from google.colab import files

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print(df.head())

print(df.shape)

print(df.columns)

print(df.isnull().sum())

print(df.duplicated().sum())

df = df.drop_duplicates()

fig1 = px.pie(
    df,
    names='Attrition',
    title='Employee Attrition Distribution',
    color='Attrition',
    color_discrete_map={'Yes':'red','No':'green'}
)

fig1.show()

fig2 = px.histogram(
    df,
    x='Department',
    color='Attrition',
    barmode='group',
    title='Attrition by Department'
)

fig2.show()

fig3 = px.bar(
    df,
    x='JobRole',
    color='Attrition',
    title='Attrition by Job Role'
)

fig3.update_layout(xaxis_tickangle=-45)

fig3.show()

fig4 = px.box(
    df,
    x='Attrition',
    y='MonthlyIncome',
    color='Attrition',
    title='Monthly Income vs Attrition'
)

fig4.show()

fig5 = px.histogram(
    df,
    x='WorkLifeBalance',
    color='Attrition',
    barmode='group',
    title='Work Life Balance vs Attrition'
)

fig5.show()

fig6 = px.histogram(
    df,
    x='Age',
    color='Attrition',
    nbins=30,
    title='Age Distribution of Employees'
)

fig6.show()

numeric_df = df.select_dtypes(include=['int64', 'float64'])

corr = numeric_df.corr()

plt.figure(figsize=(18,10))

sns.heatmap(corr, cmap='coolwarm')

plt.title("Correlation Heatmap")

plt.show()

print("1. Sales department shows higher attrition.")
print("2. Employees with lower monthly income tend to leave more.")
print("3. Poor work-life balance increases attrition.")
print("4. Certain job roles have significantly higher turnover.")
print("5. Younger employees show slightly higher attrition rates.")

print("Task 3 Completed Successfully!")
```
